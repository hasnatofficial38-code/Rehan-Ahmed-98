## Compound Interest:
### Write a Python program to calculate compound interest.
Formula: A = P * (1 + r/n) ** (n*t)

Where: A= total amount P = principal amount r = annual interest rate (decimal) n = number of times interest is compounded per year t = time in years Take P, r, n, and t as inputs from the user

```python
P = float(input("Enter principal amount: "))
r = float(input("Enter annual interest rate (decimal): "))
n = int(input("Enter number of times interest is compounded per year: "))
t = float(input("Enter time in years: "))

A = P * (1 + r / n) ** (n * t)

print("Total Amount =", A)
```

## Quadratic Equation Roots:
### Write a Python program to find the roots of a quadratic equation.
 formula:   x1 = (-b + (b**2 - 4*a*c)**0.5) / (2*a) x2 = (-b - (b**2 - 4*a*c)**0.5) / (2*a)
Take a, b, and c as inputs from the user.

```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = b ** 2 - 4 * a * c

x1 = (-b + math.sqrt(d)) / (2 * a)
x2 = (-b - math.sqrt(d)) / (2 * a)

print("Root 1 =", x1)
print("Root 2 =", x2)
```

## Gravitational Force:
### Write a Python program to calculate the gravitational force between two objects.
formula: F = G * (m1 * m2) / r**2

m1, m2 are the masses of the objects r = distance between the centers of the objects Take m1, m2 and r as inputs from the user.



```python
G = 6.67430e-11

m1 = float(input("Enter mass of first object: "))
m2 = float(input("Enter mass of second object: "))
r = float(input("Enter distance between objects: "))

F = G * (m1 * m2) / r ** 2

print("Gravitational Force =", F, "N")
```

## Future Value in Savings:
### Write a Python program to calculate the future value of an investment.
formula: FV = PV * (1 + r) ** t

where: o FV= future value o PV= present value o r= annual interest rate (as a decimal) o t= time in years Take PV, r, and t as inputs from the user.



```python
PV = float(input("Enter present value: "))
r = float(input("Enter annual interest rate: "))
t = float(input("Enter time in years: "))

FV = PV * (1 + r) ** t

print("Future Value =", FV)
```

## Work Done by a Force:
### Write a Python program to calculate the work done by a force.
formula:               W = f * d * cos(theta) 
where: W = work done f= force d = distance theta = angle between force and direction of movement (in degrees) Take f, d, and θ as inputs from the user.



```python
import math

f = float(input("Enter force: "))
d = float(input("Enter distance: "))
theta = float(input("Enter angle in degrees: "))

W = f * d * math.cos(math.radians(theta))

print("Work Done =", W, "J")
```
